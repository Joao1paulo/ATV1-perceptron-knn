```python
import numpy as np
```

```python
X_treino_fraude = np.array([
    [1.5, 1.0], [2.0, 2.0], [3.5, 1.5], [3.0, 3.0],
    [6.5, 5.0], [7.0, 7.0], [8.5, 6.0], [9.0, 8.0]
])
y_treino_fraude = np.array([0, 0, 0, 0, 1, 1, 1, 1])

def treinar_perceptron(X, y, taxa_aprendizado=0.1, epocas=100):
    # Inicializa pesos e viés com 1.0 conforme requisito
    pesos = np.ones(X.shape[1])
    vies = 1.0
    
    for _ in range(epocas):
        for i in range(X.shape[0]):
            # Cálculo da combinação linear
            z = np.dot(X[i], pesos) + vies
            # Ativação degrau
            predicao = 1 if z >= 0 else 0
            # Cálculo do erro e ajuste (Regra de Rosenblatt)
            erro = y[i] - predicao
            if erro != 0:
                pesos += taxa_aprendizado * erro * X[i]
                vies += taxa_aprendizado * erro
                
    return pesos, vies

def prever_perceptron(x, pesos, vies):
    z = np.dot(x, pesos) + vies
    return 1 if z >= 0 else 0

pesos_finais, vies_final = treinar_perceptron(X_treino_fraude, y_treino_fraude, taxa_aprendizado=0.1, epocas=100)

transacao_A = np.array([5.0, 4.0])
transacao_B = np.array([7.0, 6.5])

pred_A = prever_perceptron(transacao_A, pesos_finais, vies_final)
pred_B = prever_perceptron(transacao_B, pesos_finais, vies_final)

print("--- Desafio 1: Perceptron  ---")
print(f"Pesos finais: {pesos_finais} | Viés final: {vies_final:.2f}")
print(f"Transação A {transacao_A.tolist()} -> Classe: {pred_A} ({'Suspeita' if pred_A == 1 else 'Legítima'})")
print(f"Transação B {transacao_B.tolist()} -> Classe: {pred_B} ({'Suspeita' if pred_B == 1 else 'Legítima'})")
```

```python
X_treino_churn = np.array([
    [2.0, 1.0], [3.0, 0.0], [5.0, 1.0], [6.0, 2.0],
    [15.0, 4.0], [18.0, 5.0], [20.0, 4.0], [22.0, 6.0]
])
y_treino_churn = np.array([0, 0, 0, 0, 1, 1, 1, 1])

def classificar_knn(X_treino, y_treino, ponto, k=3, metrica='euclidiana'):
    # Processamento vetorizado das distâncias
    if metrica == 'euclidiana':
        distancias = np.sqrt(np.sum((X_treino - ponto)**2, axis=1))
    elif metrica == 'manhattan':
        distancias = np.sum(np.abs(X_treino - ponto), axis=1)
    else:
        raise ValueError("Métrica inválida.")
        
    indices_vizinhos = np.argsort(distancias)[:k]
    classes_vizinhos = y_treino[indices_vizinhos]
    distancias_vizinhos = distancias[indices_vizinhos]
    
    valores, contagens = np.unique(classes_vizinhos, return_counts=True)
    classe_predita = valores[np.argmax(contagens)]
    
    return classe_predita, indices_vizinhos, distancias_vizinhos

cliente_1 = np.array([13.0, 10.0])
cliente_2 = np.array([7.0, 1.5])

clientes = [("Cliente 1", cliente_1), ("Cliente 2", cliente_2)]

print("\n--- Desafio 2: KNN ---")
for nome, dados in clientes:
    classe, indices, dists = classificar_knn(X_treino_churn, y_treino_churn, dados, k=3, metrica='euclidiana')
    rotulo = "Alto Risco" if classe == 1 else "Baixo Risco"
    
    print(f"\n{nome} {dados.tolist()}:")
    print(f"Classe Predita: {rotulo}")
    print(f"Índices dos Vizinhos: {indices.tolist()}")
    print(f"Distâncias dos Vizinhos: {np.round(dists, 2).tolist()}")
```

```python
catalogo_servidores = np.array([
    [2.0, 4.0, 50.0],    
    [4.0, 8.0, 100.0],   
    [8.0, 16.0, 250.0],  
    [16.0, 32.0, 500.0], 
    [32.0, 64.0, 1000.0],
    [64.0, 128.0, 2000.0]
])

nomes_servidores = [
    "Micro Instância Web", "Standard App Server", "Medium Backend & Cache",
    "Database Enterprise", "High Performance Computing", "Big Data & AI Training"
]

def recomendar_servidor(catalogo, perfil, k=2):

    distancias = np.sqrt(np.sum((catalogo - perfil)**2, axis=1))
    
    indices_recomendados = np.argsort(distancias)[:k]
    distancias_recomendadas = distancias[indices_recomendados]
    
    return indices_recomendados, distancias_recomendadas


# Nota: Utilizando [12.0, 28.0, 850.0] conforme vetor fornecido no enunciado
perfil_demandado = np.array([12.0, 28.0, 850.0])
k_recomendacoes = 2

indices_rec, dists_rec = recomendar_servidor(catalogo_servidores, perfil_demandado, k=k_recomendacoes)

print("\n--- Desafio 3: Recomendação Espacial ---")
print(f"Perfil Demandado: {perfil_demandado.tolist()}\n")

for i in range(k_recomendacoes):
    idx = indices_rec[i]
    nome = nomes_servidores[idx]
    specs = catalogo_servidores[idx]
    dist = dists_rec[i]
    
    print(f"{i+1}º Lugar: {nome}")
    print(f"Especificações: {specs[0]} vCPUs, {specs[1]} GB RAM, {specs[2]} GB SSD")
    print(f"Distância Métrica: {dist:.2f}\n")
```
